```python

from pathlib import Path
from ultralytics import YOLO

base = Path(
    "/Users/mathewmorris/DataScience/Crash Detection"
)
test_dir = base / "test" / "images"

# Load the trained model
model = YOLO(
    str(base / "runs/detect/train-4/weights/best.pt")
)

# Collect every test image
extensions = {".jpg", ".jpeg", ".png", ".webp"}
test_images = sorted(
    p for p in test_dir.iterdir()
    if p.is_file() and p.suffix.lower() in extensions
)

print("Total test images:", len(test_images))

# Predict on the entire test set
results = model.predict(
    source=str(test_dir),
    conf=0.25,
    batch=8,
    save=True,
    project=str(base / "test_results"),
    name="predictions",
    exist_ok=True,
    stream=True,
    verbose=False
)

# Process results without keeping all images in memory
total = 0
class_counts = {name: 0 for name in model.names.values()}

for result in results:
    total += 1
    for box in result.boxes:
        name = model.names[int(box.cls[0])]
        class_counts[name] += 1

    if total % 100 == 0:
        print(f"Processed {total}/{len(test_images)} images")

print("\nInference complete!")
print("Images processed:", total)
print("Detection counts:", class_counts)
print("Annotated images saved to:")
print(base / "test_results/predictions")
```
